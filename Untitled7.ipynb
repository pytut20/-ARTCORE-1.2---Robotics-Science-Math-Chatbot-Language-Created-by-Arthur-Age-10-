```python
# =====================================================================
#   ARTCORE 1.2 - Robotics, Science, Math & Chatbot Language
#   Created by Arthur (Age 10)
# =====================================================================
import math
import random
import ast
import matplotlib.pyplot as plt

class ArtCoreEngine:
    def __init__(self):
        self.vars = {}
        self.brain = {}
        self.robot = {"name": "None", "x": 0, "y": 0, "battery": 100, "active": False}

    def eval_expr(self, expr):
        safe_functions = {
            ast.Add: lambda a, b: a + b,
            ast.Sub: lambda a, b: a - b,
            ast.Mult: lambda a, b: a * b,
            ast.Div: lambda a, b: a / b,
            ast.USub: lambda a: -a,
            ast.UAdd: lambda a: +a
        }

        safe_names = {
            "sqrt": math.sqrt,
            "pi": math.pi,
            "sin": math.sin,
            "cos": math.cos,
            "log": math.log10,
            "fat": math.factorial
        }
        safe_names.update(self.vars)

        def _eval(node):
            if isinstance(node, ast.Constant):
                return node.value
            elif isinstance(node, ast.Name):
                if node.id in safe_names:
                    return safe_names[node.id]
                raise KeyError(f"Variável desconhecida: {node.id}")
            elif isinstance(node, ast.BinOp):
                op_type = type(node.op)
                if op_type in safe_functions:
                    return safe_functions[op_type](_eval(node.left), _eval(node.right))
            elif isinstance(node, ast.UnaryOp):
                op_type = type(node.op)
                if op_type in safe_functions:
                    return safe_functions[op_type](_eval(node.operand))
            elif isinstance(node, ast.Call):
                if isinstance(node.func, ast.Name) and node.func.id in safe_names:
                    func = safe_names[node.func.id]
                    args = [_eval(arg) for arg in node.args]
                    return func(*args)
            elif isinstance(node, ast.Compare):
                left = _eval(node.left)
                for op, comp in zip(node.ops, node.comparators):
                    right = _eval(comp)
                    if isinstance(op, ast.Gt) and not (left > right): return False
                    if isinstance(op, ast.Lt) and not (left < right): return False
                    if isinstance(op, ast.GtE) and not (left >= right): return False
                    if isinstance(op, ast.LtE) and not (left <= right): return False
                    if isinstance(op, ast.Eq) and not (left == right): return False
                    if isinstance(op, ast.NotEq) and not (left != right): return False
                return True
            raise TypeError(f"Operação não permitida: {type(node).__name__}")

        try:
            tree = ast.parse(expr, mode='eval')
            return _eval(tree.body)
        except Exception as e:
            return f"Math Error: {e}"

    def run_line(self, line):
        line = line.strip()
        if not line or line.startswith("//"):
            return

        # --- CORE & CHATBOT ---
        if line.startswith("art.say "):
            content = line[8:].strip()
            if content.startswith('"') and content.endswith('"'):
                print(content[1:-1])
            elif content in self.vars:
                print(self.vars[content])
            else:
                print(self.eval_expr(content))

        elif line.startswith("art.let "):
            rest = line[8:].strip()
            name, val = rest.split("=", 1)
            name, val = name.strip(), val.strip()
            if val.startswith('"') and val.endswith('"'):
                self.vars[name] = val[1:-1]
            else:
                self.vars[name] = self.eval_expr(val)

        elif line.startswith("art.ask "):
            rest = line[8:].strip()
            name, question = rest.split("=", 1)
            ans = input(question.strip().strip('"') + " ")
            self.vars[name.strip()] = ans

        elif line.startswith("art.brain "):
            rest = line[10:].strip()
            if "->" in rest:
                key, reply = rest.split("->", 1)
                self.brain[key.strip().strip('"').lower()] = reply.strip().strip('"')

        elif line == "art.chat":
            print(f"\n[{self.robot['name']} AI Chat Online - Type 'exit' to stop]")
            while True:
                user_msg = input("You> ").strip().lower()
                if user_msg == "exit":
                    print("Chat ended.")
                    break
                replied = False
                for key, reply in self.brain.items():
                    if key in user_msg:
                        print(f"Bot> {reply}")
                        replied = True
                        break
                if not replied:
                    print("Bot> I don't have data on that yet. Teach me with art.brain!")

        # --- MATH & SCIENCE ---
        elif line.startswith("art.calc "):
            rest = line[9:].strip()
            name, expr = rest.split("=", 1)
            expr = expr.strip()
            if expr in self.vars:
                expr = str(self.vars[expr])
            res = self.eval_expr(expr)
            self.vars[name.strip()] = res
            print(f"[MATH] {name.strip()} = {res}")

        elif line.startswith("art.plot "):
            raw_list = line[9:].strip().replace("[", "").replace("]", "")
            numbers = [float(x.strip()) for x in raw_list.split(",")]
            plt.figure(figsize=(6, 3))
            plt.plot(numbers, marker='o', color='blue')
            plt.title("ArtCore Scientific Data Plot")
            plt.grid(True)
            plt.show()

        # --- ROBOTICS & SENSORS ---
        elif line.startswith("art.robot.start "):
            r_name = line[16:].strip().strip('"')
            self.robot = {"name": r_name, "x": 0, "y": 0, "battery": 100, "active": True}
            print(f"[ROBOT] {r_name} ONLINE | Pos: (0,0) | Battery: 100%")

        elif line.startswith("art.robot.move "):
            if not self.robot["active"]:
                print("[ERROR] Start a robot first with art.robot.start!")
                return
            parts = line[15:].strip().split(" ")
            if len(parts) == 2:
                raw_dir, raw_steps = parts
                direction = raw_dir.strip('"').lower()
                steps = int(raw_steps)
                if self.robot["battery"] >= steps * 2:
                    if direction == "forward": self.robot["y"] += steps
                    elif direction == "back": self.robot["y"] -= steps
                    elif direction == "right": self.robot["x"] += steps
                    elif direction == "left": self.robot["x"] -= steps
                    self.robot["battery"] -= steps * 2
                    print(f"[{self.robot['name']}] Moved {direction} {steps}m -> Pos: ({self.robot['x']},{self.robot['y']}) | Battery: {self.robot['battery']}%")
                else:
                    print(f"[{self.robot['name']}] LOW BATTERY! Recharge needed.")
            else:
                print('Syntax Error: Use -> art.robot.move "forward" 5')

        elif line.startswith("art.sensor.temp "):
            var_name = line[16:].strip()
            temp = round(random.uniform(18.0, 42.0), 1)
            self.vars[var_name] = temp
            print(f"[SENSOR] Temperature read: {temp}°C")

        # --- LOGIC ---
        elif line.startswith("art.if "):
            rest = line[7:].strip()
            if ":" in rest:
                cond, action = rest.split(":", 1)
                if self.eval_expr(cond.strip()):
                    self.run_line(action.strip())

        else:
            print(f"[SYNTAX ERROR] Unknown command: {line}")

    def execute(self, code):
        for line in code.strip().split("\n"):
            self.run_line(line)

art = ArtCoreEngine()
print("🚀 ArtCore 1.2 Engine Ready!")
```

```python
code = """
art.robot.start "Calculadora"
art.ask conta = "Digite sua conta:"
art.calc resultado = conta
art.say resultado
"""

art.execute(code)
```
